```python
data=np.linspace(1,20,12)
data
```

```python
twodarr=data.reshape(3,4)
twodarr
```

```python
annotarr=np.array([["a00","a01","a02","a03"],
                  ["a10","a11","a12","a13"],
                  ["a20","a21","a22","a23"],])
sns.heatmap(twodarr,cmap="viridis",annot=annotarr,fmt="s")
```

```python
plt.figure(figsize=(16,9))
annot_kws={'fontsize':10,
       "fontstyle":'italic',
       'color':"red",
       'alpha':0.5,
       'rotation':"vertical",
       'verticalalignment':"center",
       'backgroundcolor':"w"
    
}
sns.heatmap(df1,cmap="viridis",annot=True,annot_kws=annot_kws)
```

```python
plt.figure(figsize=(14,9))
annot={'fontsize':8,
       "fontstyle":'italic',
       'color':"black",
       'alpha':1.0,
       'rotation':"vertical",
       'verticalalignment':"center",
       'backgroundcolor':"w"
    
}
sns.heatmap(df1,annot=True,annot_kws=annot,vmin=2,vmax=20)
```

```python
plt.figure(figsize=(16,9))
annot={'fontsize':10,
       "fontstyle":'italic',
       'color':"b",
       'alpha':0.5,
       'rotation':"vertical",
       'verticalalignment':"center",
       'backgroundcolor':"w"
    
}
sns.heatmap(df1,annot=True,cmap="viridis",annot_kws=annot,vmin=2,vmax=20,robust=True,linewidth=2,linecolor="black")
```

```python
plt.figure(figsize=(16,9))
annot={'fontsize':10,
       "fontstyle":'italic',
       'color':"b",
       'alpha':0.5,
       'rotation':"vertical",
       'verticalalignment':"center",
       'backgroundcolor':"w"
    
}
sns.heatmap(df1,annot=True,cmap="viridis",annot_kws=annot,vmin=2,vmax=20,robust=True,linewidth=4,linecolor="black",cbar=False)
```

```python
plt.figure(figsize=(16,9))
cbar={'orientation':"horizontal",
       "shrink":1,
       'extend':"both",
       'extendfrac':0.1,
       'ticks':np.arange(0,22),
       'drawedges':True,
    
    
}
sns.heatmap(df1,annot=True,cmap="viridis",annot_kws=annot,vmin=2,vmax=20,robust=True,linewidth=4,linecolor="black",cbar_kws=cbar)
```

```python
plt.figure(figsize=(16,9))
cbar={'orientation':"horizontal",
       "shrink":1,
       'extend':None,
       'extendfrac':0.1,
       'ticks':np.arange(0,22),
       'drawedges':True,
    
    
}
sns.heatmap(df1,annot=True,xticklabels=np.arange(0,15),cmap="viridis",annot_kws=annot,vmin=2,vmax=20,robust=True,linewidth=4,linecolor="black",cbar_kws=cbar)
```

```python
plt.figure(figsize=(16,9))
cbar={'orientation':"horizontal",
       "shrink":1,
       'extend':"both",
       'extendfrac':0.1,
       'ticks':np.arange(0,22),
       'drawedges':False,
    
    
}
ob=sns.heatmap(df1,annot=True,xticklabels=np.arange(0,15),
               cmap="viridis",
               annot_kws=annot,
               vmin=2,vmax=20,
               robust=True,linewidth=4,
               linecolor="black",
               cbar_kws=cbar)
ob.set(title="who is resposible for global warming",
      xlabel="year wise",
      ylabel="country name")
```
