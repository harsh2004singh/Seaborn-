```python
var1 =np.linspace(1,10,10).reshape(2,5)
```

```python
sns.heatmap(var1, cmap="PuOr")
plt.show()
```

```python
ar =np.array([["a0","a1","a2","a3","a4",],["b0","b1","b2","b3","b4",]])
```

```python
sns.heatmap(var1, cmap="PuOr",annot=ar,fmt="s")
plt.show()
```

```python
y ={"fontsize":10,"color":"r"}
sns.heatmap(var1, cmap="PuOr",annot=ar,fmt="s",annot_kws=y)
plt.show()
```

```python
sns.heatmap(var1, cmap="PuOr",linewidths=5)
plt.show()
```

```python
sns.heatmap(var1, cmap="PuOr",linewidths=5,linecolor="g")
plt.show()
```

```python
sns.heatmap(var1, cmap="PuOr",linewidths=5,cbar=False)
plt.show()
```

```python
sns.heatmap(var1, cmap="PuOr",cbar=False,xticklabels=False)
plt.show()
```

```python
sns.heatmap(var1, cmap="PuOr",cbar=False,xticklabels=False,yticklabels=False)
plt.show()
```

```python
v=sns.heatmap(var1, cmap="PuOr",cbar=False,xticklabels=False,yticklabels=False)
v.set(xlabel ="python",ylabel ="users")
plt.show()
```

```python
v=sns.heatmap(var1, cmap="PuOr",cbar=False,xticklabels=False,yticklabels=False)
v.set(xlabel ="python",ylabel ="users")
sns.set(font_scale =15)
plt.show()
```
