```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = sns.load_dataset('tips')
```

```python
data
```

```python
sns.set(style ="whitegrid")
```

```python
sns.boxplot(x ="day",y ="total_bill",data =data)
plt.show()
```

```python
sns.boxplot(x ="day",y ="tip",data =data,hue="sex")
plt.show()
```

```python
sns.boxplot(x ="day",y ="tip",data =data,hue="sex",color="g")
plt.show()
```

```python
sns.boxplot(x ="day",y ="tip",data =data,hue="sex",palette="Accent")
plt.show()
```

```python
sns.boxplot(x ="day",y ="tip",data =data,hue="sex",order=["Fri","Sat","Thur","Sun"])
plt.show()
```

```python
sns.boxplot(x ="day",y ="tip",data =data,hue="sex",showmeans=True)
plt.show()
```
