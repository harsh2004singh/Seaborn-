```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data =sns.load_dataset("tips")
```

```python
data
```

```python
sns.pairplot(data)
plt.show()
```

```python
sns.pairplot(data,hue= "sex")
plt.show()
```

```python
sns.pairplot(data,hue= "sex",vars=['total_bill','tip'])
plt.show()
```

```python
sns.pairplot(data,hue= "sex",vars=['total_bill','tip'],hue_order=["Female","Male"])
plt.show()
```

```python
sns.pairplot(data,hue= "sex",vars=['total_bill','tip'],palette="BuGn")
plt.show()
```

```python
sns.pairplot(data,hue= "sex",x_vars=['total_bill','tip'])
plt.show()
```

```python
sns.pairplot(data,hue= "sex",y_vars=['total_bill','tip'])
plt.show()
```

```python
sns.pairplot(data,hue= "sex",kind="reg")
plt.show()
```

```python
sns.pairplot(data,hue= "sex",kind="kde")
plt.show()
```

```python
sns.pairplot(data,hue= "sex",kind="kde",diag_kind="kde")
plt.show()
```

```python
sns.pairplot(data,hue= "sex",kind="hist")
plt.show()
```

```python
sns.pairplot(data,hue= "sex",kind="scatter")
plt.show()
```
