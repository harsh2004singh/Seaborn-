```python
import matplotlib.pyplot as plt
import seaborn as sns
```

# Seaborn's catplot function is a versatile tool for creating various types of categorical plots. It provides a high-level interface for drawing attractive and informative statistical graphics, making it easier to visualize and compare different features of your data.

```python
data = sns.load_dataset("tips")
```

```python
data
```

```python
sns.catplot(x="tip",y="size",data =data)
plt.show()
```

```python
sns.catplot(x="tip",y="size",data =data,hue="sex")
plt.show()
```

```python
sns.catplot(x="tip",y="size",data =data,hue="sex",kind="bar")
plt.show()
```

```python
sns.catplot(x="tip",y="size",data =data,hue="sex",kind="point")
plt.show()
```

```python
sns.catplot(x="tip",y="size",data =data,hue="sex",kind="box")
plt.show()
```

```python
sns.catplot(x="tip",y="size",data =data,hue="sex",kind="boxen")
plt.show()
```

# Facet Grid

# Facet grids in ggplot2 allow you to create multi-panel plots by splitting the data into subsets and displaying each subset in its own panel. This is particularly useful when you have two discrete variables and want to visualize all combinations of these variables.

```python
fg =sns.FacetGrid(data)
fg.map(plt.scatter,"total_bill","tip")
plt.show()
```

```python
fg =sns.FacetGrid(data,col="sex")
fg.map(plt.scatter,"total_bill","tip")
plt.show()
```

```python
fg =sns.FacetGrid(data,col="sex",hue ="day")
fg.map(plt.scatter,"total_bill","tip").add_legend()
plt.show()
```

```python
fg =sns.FacetGrid(data,col="day",hue ="sex")
fg.map(plt.scatter,"total_bill","tip").add_legend()
plt.show()
```

```python
fg =sns.FacetGrid(data,col="sex",hue ="day")
fg.map(plt.bar,"total_bill","tip").add_legend()
plt.show()
```

```python
fg =sns.FacetGrid(data,col="day",hue ="sex")
fg.map(plt.bar,"total_bill","tip").add_legend()
plt.show()
```
