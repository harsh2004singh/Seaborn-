```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = sns.load_dataset("penguins")
```

```python
data
```

```python
data['island'].value_counts()
```

```python
sns.barplot(x=data.island,y=data.bill_length_mm)
```

```python
sns.barplot(x=data.island,y=data.bill_length_mm)
```

```python
sns.barplot(x=data.island,y=data.bill_depth_mm,hue=data.sex)
plt.show()
```

```python
or_1 =["Dream","Torgersen","Biscoe island"]
sns.barplot(x=data.island,y=data.bill_depth_mm,hue=data.sex,order =or_1)
plt.show()
```

```python
sns.barplot(x=data.island,y=data.bill_depth_mm,hue=data.sex,hue_order=["Female",'Male'])
plt.show()
```

```python
sns.barplot(x=data.island,y=data.bill_depth_mm,hue=data.sex,ci= 20)
plt.show() 
# ci range 1-100
```

```python
sns.barplot(x=data.island,y=data.bill_depth_mm,hue=data.sex ,n_boot=4)
plt.show()
```

```python
sns.barplot(x=data.island,y=data.bill_depth_mm,hue=data.sex,orient='v')
plt.show()
#h ,r 
```

```python
sns.barplot(x=data.island,y=data.bill_depth_mm,color ="g")
plt.show()
```

```python
sns.barplot(x=data.island,y=data.bill_depth_mm,palette="icefire")
plt.show()
```

```python
sns.barplot(x=data.island,y=data.bill_depth_mm,palette="icefire",saturation=0.3)
plt.show()
```

```python
sns.barplot(x=data.island,y=data.bill_depth_mm,palette="icefire",saturation=0.3,errcolor="g")
plt.show()
```

```python
sns.barplot(x=data.island,y=data.bill_depth_mm,saturation=0.3,errwidth=12)
plt.show()
```

```python
sns.barplot(x=data.island,y=data.bill_depth_mm,capsize=0.1)
plt.show()
```

```python
sns.barplot(x=data.island,y=data.bill_depth_mm,dodge=True)
plt.show()
```
