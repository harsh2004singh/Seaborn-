```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
var =[1,2,3,4,5,6,7]
var1 =[1,4,3,4,9,10,7]

import pandas as pd
x = pd.DataFrame({"var":var,"var1":var1})

sns.lineplot(x ='var',y ='var1',data =x)
```

```python
data = sns.load_dataset("penguins")
```

```python
data
```

```python
sns.lineplot(x ='bill_length_mm',y ='bill_depth_mm',data =data)
plt.show()
```

```python
sns.lineplot(x ='bill_length_mm',y ='bill_depth_mm',data =data,hue="sex")
plt.show()
```

```python
sns.lineplot(x ='bill_length_mm',y ='bill_depth_mm',data =data,hue="sex",size=20)
plt.show()
```

```python
sns.lineplot(x ='bill_length_mm',y ='bill_depth_mm',data =data,hue="sex",style="sex")
plt.show()
```

```python
sns.lineplot(x ='bill_length_mm',y ='bill_depth_mm',data =data,hue="sex",palette="Accent")
plt.show()
```

```python
sns.lineplot(x ='bill_length_mm',y ='bill_depth_mm',data =data,hue="sex",markers=["o","<"])
plt.show()
```

```python
sns.lineplot(x ='bill_length_mm',y ='bill_depth_mm',data =data,hue="sex",dashes=False)
plt.show()
```
