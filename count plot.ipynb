```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data =sns.load_dataset("tips")
```

```python
data
```

```python
sns.countplot(x ="sex",data =data)
plt.show()
```

```python
sns.countplot(x ="sex",hue ="smoker",data =data)
plt.show()
```

```python
sns.countplot(y ="sex",hue ="smoker",data =data)
plt.show()
```

```python
sns.countplot(x ="sex",hue ="smoker",data =data, palette="Accent")
plt.show()
```

```python
sns.countplot(x ="sex",hue ="smoker",data =data,color="r")
plt.show()
```

```python
sns.countplot(x ="sex",hue ="smoker",data =data,color="r",saturation=0.4)
plt.show()
```
